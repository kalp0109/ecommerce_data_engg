```python
import pyspark.sql.functions as F
```

```python
silver_autoloader = (spark.table("ecommerce.raw.brz_orders_autoloader"))
```

```python
silver_autoloader.printSchema()
```

```python
silver_autoloader =  silver_autoloader.dropDuplicates(['order_id'])
```

### Writing it to silver table

```python
(silver_autoloader.write
                .mode('overwrite')
                .option('mergeSchema', 'true')
                .format('delta')
                .saveAsTable('ecommerce.raw.slv_orders_autoloader'))
```
