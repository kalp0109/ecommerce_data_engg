```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    DoubleType
)
```

```python
cdc_schema = StructType([
    StructField("event_id", StringType(), True),
    StructField("event_ts", StringType(), True),
    StructField("order_id", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("product_id", StringType(), True),
    StructField("quantity", IntegerType(), True),
    StructField("unit_price", DoubleType(), True),
    StructField("status", StringType(), True),
    StructField("operation", StringType(), True)
])
```

```python
cdc_raw = (spark.readStream
        .format("kafka")
        .option(
            "kafka.bootstrap.servers",
            BOOTSTRAP_SERVER
        )
        .option("subscribe", "ORDER_TOPICS")
        .option("startingOffsets", "earliest")
        .option("kafka.security.protocol", "SASL_SSL")
        .option("kafka.sasl.mechanism", "PLAIN")
        .option(
            "kafka.sasl.jaas.config",
            f'kafkashaded.org.apache.kafka.common.security.plain.PlainLoginModule required username="{API_KEY}" password="{API_SECRET}";'
        )
        .load())
```

### Parse the CDC JSON

```python
from pyspark.sql import functions as F
```

```python
cdc_parsed = (
    cdc_raw
    .selectExpr("CAST(value AS STRING) AS json_value")
    .select(
        F.from_json("json_value", cdc_schema).alias("data")
    )
    .select("data.*")
    .withColumn("event_ts", F.to_timestamp("event_ts"))
)
```

### Write CDC events to Bronze

```python
cdc_bronze_query  = ( cdc_parsed.writeStream
    .format("delta")
    .outputMode("append")
    .option("checkpointLocation",  "/Volumes/ecommerce/raw/raw_data/checkpoints/brz_order_cdc")
    .trigger(availableNow=True)
    .toTable('ecommerce.raw.brz_orders_cdc')

)
```
